```python
#Parameters are like variables that will receive values when you call the function.
#It’s what gets assigned to the parameter.
#When you call a function and specify the parameter name = value directly.
def addnum(x,y):
    result=x+y
    return result 
addnum(10,20)
#x,y=parameter
#10,20 =positional arguments.x=10,y=20
```

```python
#different argument :positional,keyword,default,variable length argument
def status(job,income):
    print(job)
    print(income)
status('data scientist','200000')
#data scientist,200000=keywrod argument
```

```python
def status(job,income):
    print('income=',income)
    print('job= ',job)
status(job='data scientist',income='200000')
#doesn't maintain the order of parameter value
```

```python
def status(job,income='200000'):
    print(job)
    print(income)
status('data scientist')
#income's value is default argument
```

```python
def add_num(*x):
    print(x)
    result=sum(x)
    return result
add_num(10,20,30)
#x represents a tuple
```

```python
def add_num(x,*y):
    print(x)
    print(y)
add_num(10,20,30)
#y represents a tuple
```

```python
def add_num(x,**y):
    print(x)
    print(y)
add_num(10,aru='20',aruf='30')
#y represents a dictionary 
```
